```python
from numba import njit, prange, threading_layer
import numpy as np

@njit(parallel=True)
def random_array_fixed(n, base_seed):
    out = np.empty(n)
    for i in prange(n):
        np.random.seed(base_seed + i)  # seed depends on index
        out[i] = np.random.normal(loc=0, scale=1)
    return out
```

```python
import math

def SHOD(log10_Mh, pcen):

    """
    Standard HOD model (Zheng et al. 2007).

    Parameters:
    -----------
    - log10_Mh: float
        Logarithm (base 10) of halo mass.
    - Ac: float
        Normalization amplitude.
    - Mc: float
        Halo mass cut (log10).
    - sig_M: float
        Stepness of the step function.
    Returns:
    --------
    - float
        Expected number of galaxies in a halo of mass log10_Mh.

    """
    Ac, Mc, sig_M = pcen
    return Ac * 0.5 * (1 + erf((log10_Mh - Mc) / sig_M))
```

```python
def test_ngal(log10_Mh, fun_cHOD, fun_sHOD, Nthread, p_cen, p_sat=None, conformity=False, link_sat_to_central=False):

    nbinsM, logbinM = np.histogram(log10_Mh, bins=100)[:2]
    dM = np.diff(logbinM)[0]

    #p_sat=None
    if p_sat is not None :
        LogM = (logbinM[1:]+logbinM[:-1])*0.5
        Ncent = fun_cHOD(LogM, p_cen)
        if (Ncent < 0).any():
            Ncent[Ncent < 0] = 0
        ngal_c = nbinsM * Ncent
        if p_sat is not None :
            N_sat = fun_sHOD(LogM, p_sat)
            if (LogM < p_sat[1]).any():
                N_sat[LogM < p_sat[1]] = 0
            ngal_sat = nbinsM * Ncent*N_sat if conformity | link_sat_to_central else nbinsM* N_sat
        else:
            return ngal_c, 0
        ngal_tot = ngal_c.sum()+ngal_sat.sum()
        return ngal_tot, ngal_sat.sum()/ngal_tot
   
```
