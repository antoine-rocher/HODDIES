```python
import math
def cmp_normalizing_constant(lam, nu, max_x=1000):
    """
    Approximate Z(lam, nu) = sum_{j=0}^max_x lam^j/(j!)^nu
    Increase max_x for better accuracy.
    """
    terms = [ (lam**j) / (math.factorial(j)**nu) for j in range(max_x+1) ]
    return sum(terms)

def cmp_pmf(x, lam, nu, Z=None, max_x=1000):
    """
    Compute the CMP PMF at x.
    If Z not given, compute it internally.
    """
    if Z is None:
        Z = cmp_normalizing_constant(lam, nu, max_x=max_x)
    
    return (lam**x) / ( (math.factorial(x)**nu) * Z )

def cmp_pmf_array(lam, nu, max_x=50):
    Z = cmp_normalizing_constant(lam, nu, max_x=500)
    xs = np.arange(max_x+1)
    pmf_vals = np.array([ cmp_pmf(x, lam, nu, Z) for x in xs ])
    return xs, pmf_vals



def cmp_sample(lam, nu, size=1, max_x=500):
    xs, pmf_vals = cmp_pmf_array(lam, nu, max_x)
    cdf = np.cumsum(pmf_vals)
    cdf /= cdf[-1]   # normalize
    
    u = np.random.rand(size)
    samples = np.searchsorted(cdf, u)
    return samples


lam = 2.5
nu = 1.2

# Approximate PMF
xs, pmf = cmp_pmf_array(lam, nu, max_x=100)

# Sample from CMP
samples = cmp_sample(lam, nu, size=1000, max_x=200)
```

```python
import numpy as np
from scipy.special import gammaln  # log(n!)

def log_factorial(x):
    return gammaln(x + 1.0)

tab_gamma_ln = np.hstack([gammaln(x) for x in range(0,1000000)])
def log_factorial(x):
    return tab_gamma_ln[x+1]

def cmp_logZ(lam, nu, max_x=1000):
    x = np.arange(max_x + 1)
    log_terms = x * np.log(lam) - nu * log_factorial(x)
    m = np.max(log_terms)          # log-sum-exp trick
    return m + np.log(np.sum(np.exp(log_terms - m)))

def cmp_pmf_vec(x, lam, nu, logZ=None, max_x=1000):
    x = np.asarray(x)
    if logZ is None:
        logZ = cmp_logZ(lam, nu, max_x)

    log_p = x * np.log(lam) - nu * log_factorial(x) - logZ
    return np.exp(log_p)


def cmp_pmf_array(lam, nu, max_x=1000):
    x = np.arange(max_x + 1)
    logZ = cmp_logZ(lam, nu, max_x=max_x*2)
    pmf = cmp_pmf_vec(x, lam, nu, logZ)
    return x, pmf

    
def cmp_sample(lam, nu, size=1, max_x=10000): 
    x = np.arange(max_x + 1) 
    logZ = cmp_logZ(lam, nu, max_x=max_x*2) 
    pmf = cmp_pmf_vec(x, lam, nu, logZ) 
    cdf = np.cumsum(pmf) 
    cdf /= cdf[-1] 
    u = np.random.rand(size) 
    return np.searchsorted(cdf, u)

def cmp_mean_var_ultra(lam, nu, c=2.5):
    lam = np.asarray(lam, dtype=np.float64)
    
    # ---------- Series (λ small) ----------
    lam2 = lam*lam
    lam3 = lam2*lam
    
    a2 = 2**nu
    a3 = 6**nu
    
    Z = 1 + lam + lam2/a2 + lam3/a3
    N1 = lam + 2*lam2/a2 + 3*lam3/a3
    N2 = lam + 4*lam2/a2 + 9*lam3/a3
    
    mu_series = N1 / Z
    EX2 = N2 / Z
    var_series = EX2 - mu_series**2
    
    # ---------- Improved asymptotic ----------
    x = lam**(1/nu)
    invx = x**(-1)
    
    mu_asymp2 = (
        x
        - (nu-1)/(2*nu)
        + (nu**2 - 1)/(24*nu**2) * invx
    )
    
    var_asymp2 = (
        (x/nu) *
        (1 - (nu**2 - 1)/(12*nu**2) * invx)
    )
    
    # ---------- Adaptive blending ----------
    w = 1.0 / (1.0 + np.exp(-c*(x - 2.0)))
    
    mu = (1-w)*mu_series + w*mu_asymp2
    var = (1-w)*var_series + w*var_asymp2
    
    return mu, var

def cmp_lambda_from_mean(mu, nu):
    """Rescale λ so CMP mean ≈ mu (Poisson mean)."""
    return (mu + (nu - 1) / (2 * nu)) ** nu

def cmp_var_approx(lam, nu):
    var = lam**(1/nu) / nu
    return var


def cmp_mean_var_from_mu(mu, nu):
    lam = cmp_lambda_from_mean(mu, nu)
    var = cmp_var_approx(lam, nu)
    return lam, var

def cmp_mean_small_lambda(lam, nu):
    return lam + lam**2 * (2**(1-nu) - 1)
def cmp_var_small_lambda(lam, nu):
    return lam + lam**2 * (2**(2-nu) - 2)

def cmp_mean_var_uniform(lam, nu):
    lamrs = lam**(1/nu) - (nu-1)/(2*nu)*(1 - np.exp(-lam))
    var = lamrs / nu
    return lamrs, var




    
def cmp_mean_var_from_mu_allscale(lam, nu):
    if lam < .15:
        lamrs = cmp_mean_small_lambda(lam, nu)
        var = cmp_var_small_lambda(lam, nu) 
    elif lam < .3:
        lamrs, var = cmp_mean_var_ultra(lam, nu)
    elif lam <= .6:
        lamrs, var = cmp_mean_var_uniform(lam, nu)    
    elif lam > 0.6:
        lamrs, var = cmp_mean_var_from_mu(lam, nu)
    return lamrs, var

lam, nu = 3.0, 1.5

x, pmf = cmp_pmf_array(lam, nu, max_x=500)
samples = cmp_sample(lam, nu, size=10000)
```

```python
from HODDIES.HOD_models import _SHOD, _Nsat_pow_law
from scipy.special import gammaln  # log(n!)
import numpy as np
from numba import njit, numba
import os
import scipy


tab_gamma_ln = np.hstack([gammaln(x) for x in range(0,1000000)])
@njit(fastmath=True)
def log_factorial(x):
    return tab_gamma_ln[x+1]


@njit(fastmath=True)
def cmp_logZ(lam, nu, max_x=1000):
    x = np.arange(max_x + 1)
    log_terms = x * np.log(lam) - nu * log_factorial(x)
    m = np.max(log_terms)          # log-sum-exp trick
    return m + np.log(np.sum(np.exp(log_terms - m)))

@njit(fastmath=True)
def cmp_pmf_vec(x, lam, nu, logZ=None, max_x=1000):
    x = np.asarray(x)
    if logZ is None:
        logZ = cmp_logZ(lam, nu, max_x)

    log_p = x * np.log(lam) - nu * log_factorial(x) - logZ
    return np.exp(log_p)

@njit(fastmath=True)
def cmp_sample_v1(lam, nu, size=1): 
    max_x = max(100, int(np.round(lam)*2))
    x = np.arange(max_x*2 + 1) 
    logZ = cmp_logZ(lam, nu, max_x=max_x) 
    pmf = cmp_pmf_vec(x, lam, nu, logZ) 
    cdf = np.cumsum(pmf) 
    cdf /= cdf[-1] 
    u = np.random.rand(size) 
    # return 0
    return (cdf < u).sum()

@njit(fastmath=True)
def cmp_sample(lam, nu, size=1, max_x=11000): 
    x = np.arange(max_x + 1) 
    logZ = cmp_logZ(lam, nu, max_x=max_x) 
    pmf = cmp_pmf_vec(x, lam, nu, logZ) 
    cdf = np.cumsum(pmf) 
    cdf /= cdf[-1] 
    u = np.random.rand(size) 
    return (cdf < u).sum()

@njit(fastmath=True)
def cmp_mean_var_ultra(lam, nu, c=2.5):
    lam = np.asarray(lam, dtype=np.float64)
    
    # ---------- Series (λ small) ----------
    lam2 = lam*lam
    lam3 = lam2*lam
    
    a2 = 2**nu
    a3 = 6**nu
    
    Z = 1 + lam + lam2/a2 + lam3/a3
    N1 = lam + 2*lam2/a2 + 3*lam3/a3
    N2 = lam + 4*lam2/a2 + 9*lam3/a3
    
    mu_series = N1 / Z
    EX2 = N2 / Z
    var_series = EX2 - mu_series**2
    
    # ---------- Improved asymptotic ----------
    x = lam**(1/nu)
    invx = x**(-1)
    
    mu_asymp2 = (
        x
        - (nu-1)/(2*nu)
        + (nu**2 - 1)/(24*nu**2) * invx
    )
    
    var_asymp2 = (
        (x/nu) *
        (1 - (nu**2 - 1)/(12*nu**2) * invx)
    )
    
    # ---------- Adaptive blending ----------
    w = 1.0 / (1.0 + np.exp(-c*(x - 2.0)))
    
    mu = (1-w)*mu_series + w*mu_asymp2
    var = (1-w)*var_series + w*var_asymp2
    
    return mu, var

@njit(fastmath=True)
def cmp_lambda_from_mean(mu, nu):
    """Rescale λ so CMP mean ≈ mu (Poisson mean)."""
    return (mu + (nu - 1) / (2 * nu)) ** nu

@njit(fastmath=True)
def cmp_var_approx(lam, nu):
    var = lam**(1/nu) / nu
    return var

@njit(fastmath=True)
def cmp_mean_var_from_mu(mu, nu):
    lam = cmp_lambda_from_mean(mu, nu)
    var = cmp_var_approx(lam, nu)
    return lam, var

@njit(fastmath=True)
def cmp_mean_small_lambda(lam, nu):
    return lam + lam**2 * (2**(1-nu) - 1)

@njit(fastmath=True)
def cmp_var_small_lambda(lam, nu):
    return lam + lam**2 * (2**(2-nu) - 2)

@njit(fastmath=True)
def cmp_mean_var_uniform(lam, nu):
    lamrs = lam**(1/nu) - (nu-1)/(2*nu)*(1 - np.exp(-lam))
    var = lamrs / nu
    return lamrs, var


@njit(fastmath=True) 
def cmp_mean_var_from_mu_allscale(lam, nu):
    if lam < .15:
        lamrs = cmp_mean_small_lambda(lam, nu)
        var = cmp_var_small_lambda(lam, nu) 
    elif lam < .3:
        lamrs, var = cmp_mean_var_ultra(lam, nu)
    elif lam <= .6:
        lamrs, var = cmp_mean_var_uniform(lam, nu)    
    elif lam > 0.6:
        lamrs, var = cmp_mean_var_from_mu(lam, nu)
    return lamrs, var

@njit(parallel=True, fastmath=True)
def compute_N(log10_Mh, fun_cHOD, fun_sHOD, p_cen, p_sat, nu, p_ab=None, Nthread=32, ab_arr=None, conformity=False, link_sat_to_central=False, seed=None):
    """
    Compute the number of central and satellite galaxies in a halo using a HOD (Halo Occupation Distribution) model.

    Parameters:
        log10_Mh : ndarray
            Logarithmic mass of halos.
        fun_cHOD : callable
            Function to compute central occupation based on halo mass.
        fun_sHOD : callable
            Function to compute satellite occupation based on halo mass.
        p_cen : ndarray
            Parameters for the central HOD function.
        p_sat : ndarray
            Parameters for the satellite HOD function.
        p_ab : ndarray or None, optional
            Assembly bias parameters (central and satellite). Default is None.
        Nthread : int, optional
            Number of threads to use in parallel computation. Default is 32.
        ab_arr : ndarray or None, optional
            Assembly bias array per halo. Default is None.
        conformity : bool, optional
            Whether satellite number is correlated with central galaxy presence. Default is False.
        seed : ndarray or None, optional
            Seed array for RNG per thread. Default is None.

    Returns:
        Ncent : ndarray
            Expected number of central galaxies.
        N_sat : ndarray
            Expected number of satellite galaxies.
        cond_cent : ndarray
            Boolean array indicating presence of a central galaxy.
        proba_sat : ndarray
            Sampled number of satellites per halo.
    """

    
    numba.set_num_threads(Nthread)
    # starting index of each thread
    hstart = np.rint(np.linspace(0, len(log10_Mh), Nthread + 1))
    cond_cent = np.zeros(log10_Mh.size, dtype=np.bool_)
    proba_sat = np.empty_like(log10_Mh, dtype=np.int64)
    Nb_sat = 0
    N_sat = np.empty_like(log10_Mh)

    # figuring out the number of halos kept for each thread
    for tid in numba.prange(Nthread):
        if seed is not None:
            np.random.seed(seed[tid])
        for i in range(int(hstart[tid]), int(hstart[tid + 1])):
            M_0 = p_sat[1]
            Ncent = fun_cHOD(log10_Mh[i], p_cen)
            if p_ab is not None:
                Ncent *= (1 + np.sum(p_ab[0] * ab_arr[i])*(1-Ncent))
            cond_cent[i] = Ncent - np.random.uniform(0, 1) > 0
            if p_sat is not None :
                M_0 = p_sat[1]
                if log10_Mh[i] <= M_0:
                    N_sat[i] = 0
                elif np.abs(log10_Mh[i]- M_0) < -0.001:
                    p_sat[1] += 0.001
                    N_sat[i] = fun_sHOD(log10_Mh[i], p_sat)
                    p_sat[1] -= 0.001
                else:
                    N_sat[i] = fun_sHOD(log10_Mh[i], p_sat)
                if link_sat_to_central:
                    N_sat[i] *= Ncent
                if p_ab is not None:
                        N_sat[i] *= (1 + np.sum(p_ab[1] * ab_arr[i]))
                if N_sat[i] <=0:
                    proba_sat[i] = 0
                elif conformity:
                    proba_sat[i] = np.random.poisson(N_sat[i]*cond_cent[i])
                else :
                    # proba_sat[i] = np.random.poisson(N_sat[i])
                    lam_cmp, vaar = cmp_mean_var_from_mu_allscale(N_sat[i], nu)
                    proba_sat[i] = cmp_sample_v1(lam_cmp, nu)
                if proba_sat[i] > 0:
                    Nb_sat += proba_sat[i]
                    
    return N_sat, proba_sat, Nb_sat
log10_Mh = np.geomspace(11,14, 1000, dtype=np.float32)
p_cen = np.array([1., 12.5, 0.2], dtype=np.float32)
p_sat = np.array([1., 12, 13, 1], dtype=np.float32)
nu = 1.5
```

```python
@njit(fastmath=True)
def cmp_lambda_small_mu(mu, nu):
    """
    Inverse of small-lambda CMP mean expansion
    up to O(mu^4).
    """

    a2 = 2.0**(-nu)
    a3 = 6.0**(-nu)
    a4 = 24.0**(-nu)

    A = 2*a2 - 1
    B = 3*a3 - 3*a2 + 1
    C = 4*a4 - 4*a3 - 2*a2*a2 + 4*a2 - 1

    return (
        mu
        - A*mu**2
        # + (2*A*A - B)*mu**3
        # + (-5*A**3 + 5*A*B - C)*mu**4
    )
```

```python
import numpy as np
import math
from numba import njit, prange

# -------------------------------------------------
# Utilities
# -------------------------------------------------

# @njit(fastmath=True)
# def log_factorial_numba(x):
#     return math.lgamma(x + 1.0)

tab_gamma_ln = np.hstack([gammaln(x) for x in range(0,100000)])
@njit(fastmath=True)
def log_factorial_numba(x):
    return tab_gamma_ln[x]


# -------------------------------------------------
# log Z using log-sum-exp
# -------------------------------------------------

@njit(fastmath=True)
def cmp_logZ_numba(lam, nu, max_x):
    m = -1e300

    # find max log term
    for x in range(max_x + 1):
        val = x * np.log(lam) - nu * log_factorial_numba(x)
        if val > m:
            m = val

    # log-sum-exp
    s = 0.0
    for x in range(max_x + 1):
        val = x * np.log(lam) - nu * log_factorial_numba(x)
        s += np.exp(val - m)

    return m + np.log(s)

# -------------------------------------------------
# PMF
# -------------------------------------------------

@njit(fastmath=True)
def cmp_pmf_array_numba(lam, nu, max_x):
    logZ = cmp_logZ_numba(lam, nu, max_x)

    pmf = np.empty(max_x + 1)
    for x in range(max_x + 1):
        log_p = x * np.log(lam) - nu * log_factorial_numba(x) - logZ
        pmf[x] = np.exp(log_p)

    return pmf


# -------------------------------------------------
# Sampling
# -------------------------------------------------

@njit(fastmath=True)
def cmp_sample_numba(lam, nu, size, max_x):
    pmf = cmp_pmf_array_numba(lam, nu, max_x)

    # build CDF
    cdf = np.empty_like(pmf)
    cumsum = 0.0
    for i in range(len(pmf)):
        cumsum += pmf[i]
        cdf[i] = cumsum

    # normalize
    total = cdf[-1]
    for i in range(len(cdf)):
        cdf[i] /= total

    # sample
    out = np.empty(size, dtype=np.int64)

    for i in range(size):
        u = np.random.random()
        for j in range(len(cdf)):
            if u <= cdf[j]:
                out[i] = j
                break

    return out

@njit(fastmath=True)
def cmp_sample_numba_single(lam, nu, max_x):
    pmf = cmp_pmf_array_numba(lam, nu, max_x)

    # build CDF
    cdf = np.empty_like(pmf)
    cumsum = 0.0
    for i in range(len(pmf)):
        cumsum += pmf[i]
        cdf[i] = cumsum

    # normalize
    total = cdf[-1]
    for i in range(len(cdf)):
        cdf[i] /= total

    u = np.random.uniform()
    for j in range(len(cdf)):
        if u <= cdf[j]:
            out = j
            break

    return out


# -------------------------------------------------
# Parallel batch sampling
# -------------------------------------------------

@njit(parallel=True, fastmath=True)
def cmp_sample_grid(lams, nu, size, max_x):
    n = len(lams)
    means = np.zeros(n)
    vars_ = np.zeros(n)

    for i in prange(n):
        samples = cmp_sample_numba(lams[i], nu, size, max_x)
        m = 0.0
        for s in samples:
            m += s
        m /= size

        v = 0.0
        for s in samples:
            diff = s - m
            v += diff * diff
        v /= size

        means[i] = m
        vars_[i] = v

    return means, vars_


@njit(parallel=True, fastmath=True)
def cmp_mean_var_from_mu_allscale_numba(lams, nu):
    """
        Parallel over Nsat
    """
    n = len(lams)
    mu = np.empty(n)
    var = np.empty(n)

    for i in prange(n):
        lam = lams[i]

        if lam < 0.15:
            mu_i = lam + lam**2 * (2**(1-nu) - 1)
            var_i = lam + lam**2 * (2**(1-nu) - 2)

        elif lam < 0.3:
            lam2 = lam*lam
            lam3 = lam2*lam

            a2 = 2**nu
            a3 = 6**nu

            Z = 1 + lam + lam2/a2 + lam3/a3
            N1 = lam + 2*lam2/a2 + 3*lam3/a3
            N2 = lam + 4*lam2/a2 + 9*lam3/a3

            mu_i = N1 / Z
            var_i = N2 / Z - mu_i**2

        elif lam <= 0.6:
            mu_i = lam**(1/nu) - (nu-1)/(2*nu)*(1 - math.exp(-lam))
            var_i = mu_i / nu

        else:
            lamrs = (lam + (nu - 1) / (2 * nu)) ** nu
            mu_i = lamrs
            var_i = lamrs / nu

        mu[i] = mu_i
        var[i] = var_i

    return mu, var


```

```python
from numba import njit

@njit(inline='always')
def searchsorted_numba(a, v):
    lo = 0
    hi = a.shape[0]

    while lo < hi:
        mid = (lo + hi) >> 1
        if a[mid] < v:
            lo = mid + 1
        else:
            hi = mid
    return lo
```

```python
import numpy as np

def cmp_mean_var_global(lam, nu, c=3.0):
    lam = np.asarray(lam, dtype=np.float64)
    
    # ----- Series approximation -----
    lam2 = lam*lam
    lam3 = lam2*lam
    
    a2 = 2**nu
    a3 = 6**nu
    
    Z = 1 + lam + lam2/a2 + lam3/a3
    N1 = lam + 2*lam2/a2 + 3*lam3/a3
    N2 = lam + 4*lam2/a2 + 9*lam3/a3
    
    mu_series = N1 / Z
    EX2 = N2 / Z
    var_series = EX2 - mu_series**2
    
    # ----- Asymptotic approximation -----
    mu_asymp = lam**(1/nu) - (nu-1)/(2*nu)
    var_asymp = mu_asymp / nu
    
    # ----- Smooth blending weight -----
    w = 1.0 / (1.0 + np.exp(-c*(lam - 1.0)))
    
    mu = (1-w)*mu_series + w*mu_asymp
    var = (1-w)*var_series + w*var_asymp
    
    return mu, var

import numpy as np




def cmp_mean_var_ultra(lam, nu, c=2.5):
    lam = np.asarray(lam, dtype=np.float64)
    
    # ---------- Series (λ small) ----------
    lam2 = lam*lam
    lam3 = lam2*lam
    
    a2 = 2**nu
    a3 = 6**nu
    
    Z = 1 + lam + lam2/a2 + lam3/a3
    N1 = lam + 2*lam2/a2 + 3*lam3/a3
    N2 = lam + 4*lam2/a2 + 9*lam3/a3
    
    mu_series = N1 / Z
    EX2 = N2 / Z
    var_series = EX2 - mu_series**2
    
    # ---------- Improved asymptotic ----------
    x = lam**(1/nu)
    invx = x**(-1)
    
    mu_asymp2 = (
        x
        - (nu-1)/(2*nu)
        + (nu**2 - 1)/(24*nu**2) * invx
    )
    
    var_asymp2 = (
        (x/nu) *
        (1 - (nu**2 - 1)/(12*nu**2) * invx)
    )
    
    # ---------- Adaptive blending ----------
    w = 1.0 / (1.0 + np.exp(-c*(x - 2.0)))
    
    mu = (1-w)*mu_series + w*mu_asymp2
    var = (1-w)*var_series + w*var_asymp2
    
    return mu, var

nus=np.arange(0.5, 1, 0.1)
lam = np.geomspace(.01, 0.15, 50)

for i, nu in enumerate(nus):
    lamrs = cmp_mean_small_lambda(lam, nu)
    mean_samples, var_samples = np.vstack([((ttt.mean(), ttt.std()**2)) for ttt in [cmp_sample(ll, nu, size=50000, max_x=5000) for ll in lamrs]]).T
    # plt.plot(lam, mean, lw=2, ls='-', c='gray')
    plt.plot(lam, (mean_samples-lam), lw=0.8, ls='-', c='g', alpha=np.linspace(0.1, 1, len(nus))[i])
    # plt.plot(lam, lam-var_samples, lw=0.8, ls='--', c='r', alpha=np.linspace(0.1, 1, len(nus))[int((nu-1)/(1.5-1)*(len(nus)-1))])



lam = np.geomspace(.15, 0.3, 50)

for i, nu in enumerate(nus):
    lamrs, var = cmp_mean_var_ultra(lam, nu)
    mean_samples, var_samples = np.vstack([((ttt.mean(), ttt.std()**2)) for ttt in [cmp_sample(ll, nu, size=50000, max_x=5000) for ll in lamrs]]).T
    # plt.plot(lam, mean, lw=2, ls='-', c='gray')
    plt.plot(lam, (mean_samples-lam), lw=0.8, ls='-', c='r', alpha=np.linspace(0.1, 1, len(nus))[i], label=f'nu={nu:.1f}')

lam = np.geomspace(.3, 0.7, 20)

for i, nu in enumerate(nus):
    lamrs = cmp_mean_uniform(lam, nu)
    mean_samples, var_samples = np.vstack([((ttt.mean(), ttt.std()**2)) for ttt in [cmp_sample(ll, nu, size=50000, max_x=5000) for ll in lamrs*1.02]]).T
    # plt.plot(lam, mean, lw=2, ls='-', c='gray')
    plt.plot(lam, (mean_samples-lam), lw=0.8, ls='-', c='purple', alpha=np.linspace(0.1, 1, len(nus))[i], label=f'nu={nu:.1f}')

lam = np.geomspace(.5, 3, 25)
for i, nu in enumerate(nus):
    lamrs, mean, var = cmp_mean_var_from_mu(lam, nu)
    mean_samples, var_samples = np.vstack([((ttt.mean(), ttt.std()**2)) for ttt in [cmp_sample(ll, nu, size=50000, max_x=5000) for ll in lamrs]]).T
    # plt.plot(lam, mean, lw=2, ls='-', c='gray')
    plt.plot(lam, (mean_samples-lam), lw=0.8, ls='-', c='b', alpha=np.linspace(0.1, 1, len(nus))[i], label=f'nu={nu:.1f}')


plt.axline((0, 0), slope=0, c='k', lw=1, ls='--')
plt.fill_between(np.linspace(0.1,1), -0.05, 0.05, color='gray', alpha=0.3, label='Marge ±0.05')
# plt.axline((0, 10), slope=1, c='k', lw=1, ls='--')
plt.xscale('log')
# plt.yscale('log')
plt.ylim(-0.2,0.2)
plt.xlabel('Predicted CMP')
plt.ylabel('Sampled CMP')
```

```python
import numpy as np

def cmp_exact_moments(lam, nu, kmax=200, tol=1e-14):
    ks = np.arange(kmax)
    
    # log form for numerical stability
    log_terms = ks*np.log(lam) - nu*np.log(np.maximum(1, np.cumprod(np.arange(0, kmax))))
    
    # Better: use gammaln
    from scipy.special import gammaln
    log_terms = ks*np.log(lam) - nu*gammaln(ks+1)
    
    terms = np.exp(log_terms - np.max(log_terms))
    Z = np.sum(terms)
    
    probs = terms / Z
    
    mu = np.sum(ks * probs)
    var = np.sum((ks - mu)**2 * probs)
    
    return mu, var


lams = np.geomspace(0.01, 10, 40)
nus = np.linspace(0.6, 1.5, 10)

for nu in nus:
    print(f"\nnu = {nu}")
    max_err_mu = 0
    max_err_var = 0
    
    for lam in lams:
        mu_exact, var_exact = cmp_exact_moments(lam, nu)
        mu_app, var_app = cmp_mean_var_global(lam, nu)
        
        err_mu = abs(mu_app - mu_exact) / mu_exact
        err_var = abs(var_app - var_exact) / var_exact
        
        max_err_mu = max(max_err_mu, err_mu)
        max_err_var = max(max_err_var, err_var)
    
    print("Max relative mean error:", max_err_mu)
    print("Max relative var error :", max_err_var)
```

```python
import numpy as np
import sympy as sp

def cmp_mean_var_allscale_v2(lam, nu, order=6):
    """
    Compute CMP mean and variance across all scales using:
    - small-lambda series (λ ≲ 0.15)
    - intermediate-lambda series (λ ≲ 0.5) up to given order
    - large-lambda asymptotic formula
    
    Parameters
    ----------
    lam : float or np.ndarray
        CMP lambda parameter (>0)
    nu : float
        CMP dispersion parameter (>0)
    order : int
        Maximum order for intermediate-lambda expansion
    
    Returns
    -------
    mu : np.ndarray
        Mean approximation
    var : np.ndarray
        Variance approximation
    """
    lam = np.atleast_1d(lam).astype(np.float64)
    mu = np.zeros_like(lam)
    var = np.zeros_like(lam)
    
    # 1️⃣ Small-lambda region: λ <= 0.15 (series expansion)
    mask_small = lam <= 0.15
    if mask_small.any():
        # Truncated series expansion up to λ^3 is sufficient
        l = lam[mask_small]
        a2 = 2**(-nu)
        a3 = 6**(-nu)
        Z = 1 + l + a2*l**2 + a3*l**3
        N1 = l + 2*a2*l**2 + 3*a3*l**3
        N2 = l + 4*a2*l**2 + 9*a3*l**3
        mu[mask_small] = N1/Z
        var[mask_small] = N2/Z - (N1/Z)**2

    # 2️⃣ Intermediate-lambda region: 0.15 < λ <= 0.5
    mask_mid = (lam > 0.15) & (lam <= 0.5)
    if mask_mid.any():
        l = lam[mask_mid]
        
        # Use symbolic expansion for mean and variance
        lam_sym = sp.symbols('lam', positive=True)
        nu_sym = sp.symbols('nu', positive=True)
        # Generate a_k = 1/(k!)^nu up to order
        a = [1] + [(sp.factorial(k))**(-nu_sym) for k in range(1, order+1)]
        Z_sym = sum(a[k]*lam_sym**k for k in range(0, order+1))
        N1_sym = sum(k*a[k]*lam_sym**k for k in range(1, order+1))
        N2_sym = sum(k*(k-1)*a[k]*lam_sym**k for k in range(2, order+1))
        
        mu_series = N1_sym/Z_sym
        F2_series = N2_sym/Z_sym
        var_series = F2_series + mu_series - mu_series**2
        
        # Lambdify for fast numerical evaluation
        mu_func = sp.lambdify(lam_sym, mu_series.subs(nu_sym, nu), 'numpy')
        var_func = sp.lambdify(lam_sym, var_series.subs(nu_sym, nu), 'numpy')
        
        mu[mask_mid] = mu_func(l)
        var[mask_mid] = var_func(l)
    
    # 3️⃣ Large-lambda region: λ > 0.5 (asymptotic expansion)
    mask_large = lam > 0.5
    if mask_large.any():
        l = lam[mask_large]
        x = l**(1/nu)
        invx = 1/x
        mu_asymp = x - (nu-1)/(2*nu) + (nu**2-1)/(24*nu**2)*invx
        var_asymp = (x/nu)*(1 - (nu**2-1)/(12*nu**2)*invx)
        mu[mask_large] = mu_asymp
        var[mask_large] = var_asymp
    
    return mu.squeeze(), var.squeeze()

nus=np.arange(0.6, 1.5, 0.1)

lam = np.geomspace(.08, 1, 100)

for i, nu in enumerate(nus):
    lamrs = [cmp_mean_var_allscale_v2(ll, nu, order=6)[0] for ll in lam]
    mean_samples, var_samples = np.vstack([((ttt.mean(), ttt.std()**2)) for ttt in [cmp_sample(ll, nu, size=50000, max_x=5000) for ll in lamrs]]).T
    # plt.plot(lam, mean, lw=2, ls='-', c='gray')
    plt.plot(lam, (mean_samples-lam)/lam, lw=0.8, ls='-', c='r', alpha=np.linspace(0.1, 1, len(nus))[i], label=f'nu={nu:.1f}')
    # plt.plot(lam, lam-var_samples, lw=0.8, ls='--', c='r', alpha=np.linspace(0.1, 1, len(nus))[int((nu-1)/(1.5-1)*(len(nus)-1))])



plt.axline((0, 0), slope=0, c='k', lw=1, ls='--')
plt.xscale('log')
# plt.yscale('log')
plt.xlabel('Lambda poisson')
plt.ylabel('Mean cmp')
```

```python
import sympy as sp

def cmp_symbolic_expansion(nu, order=6):
    """
    Generate symbolic expansion of CMP mean and second factorial moment
    up to given order in lambda.
    
    Parameters
    ----------
    nu : symbolic or numeric
        CMP dispersion parameter (nu > 0)
    order : int
        Maximum order in lambda for expansion
        
    Returns
    -------
    mu_series : sympy expression
        Mean expansion up to order `order`
    F2_series : sympy expression
        Second factorial moment expansion up to order `order`
    var_series : sympy expression
        Variance expansion up to order `order`
    """
    lam = sp.symbols('lam', positive=True)
    
    # Compute a_k = 1/(k!)^nu
    a = [1]  # a_0 = 1
    for k in range(1, order+1):
        a.append((sp.factorial(k))**(-nu))
    
    # Partition function truncated
    Z = sum(a[k]*lam**k for k in range(0, order+1))
    
    # Mean numerator
    N1 = sum(k*a[k]*lam**k for k in range(1, order+1))
    mu_series = sp.series(N1/Z, lam, 0, order+1).removeO()
    
    # Second factorial moment numerator
    N2 = sum(k*(k-1)*a[k]*lam**k for k in range(2, order+1))
    F2_series = sp.series(N2/Z, lam, 0, order+1).removeO()
    
    # Variance
    var_series = F2_series + mu_series - mu_series**2
    
    return lam, mu_series, F2_series, var_series

# Example usage
nu = sp.symbols('nu', positive=True)
lam, mu_exp, F2_exp, var_exp = cmp_symbolic_expansion(nu, order=6)

print("Mean expansion up to order 6:")
sp.pprint(mu_exp)
print("\nSecond factorial moment expansion up to order 6:")
sp.pprint(F2_exp)
print("\nVariance expansion up to order 6:")
sp.pprint(var_exp)
```

```python
def Nsat_pow_law(log10_Mh, As, M_0, M_1, alpha):

    """
    Power-law model for satellite galaxies (Zheng et al. 2005).

    Parameters:
    -----------
    - log10_Mh: float
        Logarithm (base 10) of halo mass.
    - As: float
        Normalization amplitude.
    - M_0: float
        Minimum halo mass to host a satellite galaxy.
    - M_1: float
        Characteristic halo mass to host a satellite (log10).
    - alpha: float
        Slope of the pozer law distribution.

    Returns:
    --------
    - float
        Expected number of satellite galaxies in a halo of mass log10_Mh.

    """
    
    N_sat = As * ((10**log10_Mh - 10**M_0) / 10**M_1)**alpha
    return N_sat
```

```python
def cmp_lambda_from_mean(mu, nu, tol=1e-20, maxiter=50):
    mu = np.asarray(mu, dtype=np.float64)
    
    # Initial guess
    lam = (mu + (nu-1)/(2*nu))**nu
    
    for _ in range(maxiter):
        exp_term = np.exp(-lam)
        
        f = lam**(1/nu) - (nu-1)/(2*nu)*(1-exp_term) - mu
        df = (1/nu)*lam**(1/nu-1) - (nu-1)/(2*nu)*exp_term
        
        step = f / df
        lam -= step
        
        if np.all(np.abs(step) < tol):
            break
    
    return lam
```

```python
import math

def SHOD(log10_Mh, pcen):

    """
    Standard HOD model (Zheng et al. 2007).

    Parameters:
    -----------
    - log10_Mh: float
        Logarithm (base 10) of halo mass.
    - Ac: float
        Normalization amplitude.
    - Mc: float
        Halo mass cut (log10).
    - sig_M: float
        Stepness of the step function.
    Returns:
    --------
    - float
        Expected number of galaxies in a halo of mass log10_Mh.

    """
    Ac, Mc, sig_M = pcen
    return Ac * 0.5 * (1 + erf((log10_Mh - Mc) / sig_M))
```
