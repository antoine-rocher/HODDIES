```python
import matplotlib.pyplot as plt
import torch
import torch.optim as optim
import zuko

from itertools import islice
from lampe.data import JointLoader, JointDataset
from lampe.inference import NPE, NPELoss
from lampe.plots import corner, mark_point, nice_rc
from lampe.utils import GDStep
from tqdm import trange
import numpy as np
import os
import torch
from torch import nn
import glob
from pycorr.utils import cov_to_corrcoef
from torch.utils.data import Dataset, random_split, DataLoader, TensorDataset
from typing import OrderedDict
from torch.optim.lr_scheduler import ReduceLROnPlateau


X, y = train_Dataset.extract_data()
batch_size = 256
dataset_loader = JointDataset(X, y,batch_size=batch_size, shuffle=True)
X_val, y_val = val_loader.dataset[:]
dataset_loader_val = JointDataset(X_val, y_val, batch_size=batch_size, shuffle=True)

estimator = NPE(X.shape[-1], y.shape[-1], transforms=3, hidden_features=[128] * 3).to(model.device)

loss = NPELoss(estimator)   


optimizer = optim.Adam(estimator.parameters(), lr=1e-3)
step = GDStep(optimizer, clip=0.5)  # gradient descent step with gradient clipping

# estimator.train()

# for epoch in (bar := trange(256, unit='epoch')):
#     losses = []

#     for theta, x in islice(dataset_loader, batch_size):  # 256 batches per epoch
#         losses.append(step(loss(theta, x))) 

#     bar.set_postfix(loss=torch.stack(losses).mean().item())
nepoch = 256
tot_loss_val, tot_loss_train = [], []
scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=10)


use_discriminator=True
for epoch in (bar := trange(nepoch, unit='epoch')):
    losses_train, losses_val = [], []
    estimator.train()
    for theta, x in islice(dataset_loader, batch_size):  # 256 batches per epoch
        if use_discriminator:
            theta = theta.to(model.device)
            x = x.to(model.device)
            y_emu, var_emu = model.predict(estimator.flow(x).sample((1,))[0])
            loss_yemu = nn.GaussianNLLLoss(full=True)(y_emu, x, var_emu)
        else:
            loss_yemu =0
        losses_train.append(step(loss(theta, x)+loss_yemu)) 
    estimator.eval()
    with torch.no_grad():
        for theta, x in islice(dataset_loader_val, batch_size):  # 256 batches per epoch
            if use_discriminator:
                theta = theta.to(model.device)
                x = x.to(model.device)
                y_emu, var_emu = model.predict(estimator.flow(x).sample((1,))[0])
                loss_yemu = nn.GaussianNLLLoss(full=True)(y_emu, x, var_emu)
            else:
                loss_yemu =0
            losses_val.append(loss(theta, x)+loss_yemu)
    
    tot_loss_train += [torch.stack(losses_train).mean().cpu().numpy()]
    tot_loss_val += [torch.stack(losses_val).mean().cpu().numpy()]

    scheduler.step(tot_loss_val[-1])
    bar.set_postfix(train_loss=tot_loss_train[-1], val_loss=tot_loss_val[-1])
```

```python
"""
NPE training with a frozen emulator as a self-consistency ("discriminator") term.

The idea: sample theta ~ q(theta | x) from the flow, push it through the
pre-trained emulator, and penalise the flow when the predicted data vector
does not reproduce x.

FIXES relative to the original loop
-----------------------------------
1. `.sample()` -> `.rsample()`.  torch.distributions.Distribution.sample() is
   wrapped in torch.no_grad(), so the emulator term had NO gradient path to the
   flow parameters. It was computed, added to the loss, and did nothing. This
   is the one change that makes the whole idea function.

2. theta/x moved to the device unconditionally. In the original they were moved
   only inside `if use_discriminator`, so `use_discriminator=False` would feed
   CPU tensors to a CUDA estimator.

3. The emulator is explicitly frozen and put in eval mode. The optimizer only
   holds estimator.parameters() so its weights were never updated, but dropout
   and batchnorm would still have been in training mode, injecting noise into
   the penalty.

4. A weight `lambda_emu` on the penalty, and a warmup over the first epochs.
   Applying it from step 0 against an untrained flow produces garbage samples
   and a large, uninformative gradient.

5. `n_batches` separated from `batch_size`. The original used
   `islice(loader, batch_size)`, which only did what the comment said because
   both happened to be 256.

6. Validation reports both terms, so the scheduler is monitoring the same
   objective that is being trained.

A CAVEAT WORTH TAKING SERIOUSLY
------------------------------
E_q[-log p(x|theta)] is mode-seeking. On its own it is minimised by collapsing
q onto the best-fit theta, so this term shrinks the posterior and makes it
overconfident. The NPE loss pulls the other way, and `lambda_emu` sets the
balance -- but the calibration guarantee of the NPE proper scoring rule is
lost once lambda_emu > 0.

The principled version adds the entropy of q, turning the penalty into a
reverse-KL variational objective:

    E_q[ -log p(x|theta) - log p(theta) + log q(theta|x) ]

Set `use_entropy=True` to include the +log q term, which counteracts the
collapse. Either way, check the posterior width against a held-out truth with
simulation-based calibration before trusting the result.
"""

import numpy as np
import torch
import torch.optim as optim
from torch import nn
from torch.optim.lr_scheduler import ReduceLROnPlateau
from itertools import islice
from tqdm import trange

from lampe.data import JointDataset
from lampe.inference import NPE, NPELoss
from lampe.utils import GDStep


# --------------------------------------------------------------------------
# Data
# --------------------------------------------------------------------------
X, y = train_Dataset.extract_data()                      # noqa: F821  X=theta, y=x
X_val, y_val = val_loader.dataset[:]                     # noqa: F821

batch_size = 256
n_batches  = 256          # batches per epoch, NOT the same thing as batch_size

trainset = JointDataset(X, y, batch_size=batch_size, shuffle=True)
valset   = JointDataset(X_val, y_val, batch_size=batch_size, shuffle=False)

device = model.device                                    # noqa: F821


# --------------------------------------------------------------------------
# Estimator
# --------------------------------------------------------------------------
estimator = NPE(X.shape[-1], y.shape[-1],
                transforms=3, hidden_features=[256] * 3).to(device)

npe_loss  = NPELoss(estimator)
optimizer = optim.Adam(estimator.parameters(), lr=1e-3)
step      = GDStep(optimizer, clip=0.5)
scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=10)


# --------------------------------------------------------------------------
# Frozen emulator
# --------------------------------------------------------------------------
use_discriminator = True
use_entropy       = False     # True -> reverse-KL form, resists mode collapse
lambda_emu        = 0.5       # penalty weight; start small
warmup_epochs     = 20        # ramp lambda_emu in linearly over these epochs
n_emu_samples     = 1         # posterior samples per observation for the penalty

if use_discriminator:
    model.eval()                                          # noqa: F821
    for p in model.parameters():                          # noqa: F821
        p.requires_grad_(False)

gauss_nll = nn.GaussianNLLLoss(full=True, reduction='mean')


def emulator_penalty(x, n_samples=1, with_entropy=False):
    """
    E_{theta ~ q(.|x)} [ -log p(x | theta) ]  (+ log q if with_entropy)

    rsample keeps the reparametrization path open so the gradient reaches the
    flow parameters. Using sample() here silently zeroes this whole term.
    """
    posterior = estimator.flow(x)
    theta_s   = posterior.rsample((n_samples,))           # (S, B, n_dim)

    flat   = theta_s.reshape(-1, theta_s.shape[-1])
    y_emu, var_emu = model.predict(flat)                  # noqa: F821
    var_emu = var_emu.clamp_min(1e-8)

    target = x.unsqueeze(0).expand(n_samples, *x.shape).reshape(-1, x.shape[-1])
    nll    = gauss_nll(y_emu, target, var_emu)

    if with_entropy:
        # + E_q[log q]: negative entropy, penalises collapse
        nll = nll + posterior.log_prob(theta_s).mean()

    return nll


# --------------------------------------------------------------------------
# Training
# --------------------------------------------------------------------------
nepoch = 1024
tot_loss_train, tot_loss_val, tot_loss_emu = [], [], []

for epoch in (bar := trange(nepoch, unit='epoch')):

    lam = lambda_emu * min(1.0, (epoch + 1) / max(warmup_epochs, 1)) \
        if use_discriminator else 0.0

    estimator.train()
    losses_train, losses_emu = [], []

    for theta, x in islice(trainset, n_batches):
        theta = theta.to(device, non_blocking=True)
        x     = x.to(device, non_blocking=True)

        l_npe = npe_loss(theta, x)

        if lam > 0:
            l_emu = emulator_penalty(x, n_emu_samples, use_entropy)
            losses_emu.append(l_emu.detach())
        else:
            l_emu = torch.zeros((), device=device)

        losses_train.append(step(l_npe + lam * l_emu))

    estimator.eval()
    losses_val = []
    with torch.no_grad():
        for theta, x in islice(valset, n_batches):
            theta = theta.to(device, non_blocking=True)
            x     = x.to(device, non_blocking=True)
            losses_val.append(npe_loss(theta, x))

    tot_loss_train.append(torch.stack(losses_train).mean().cpu().numpy())
    tot_loss_val.append(torch.stack(losses_val).mean().cpu().numpy())
    tot_loss_emu.append(torch.stack(losses_emu).mean().cpu().numpy()
                        if losses_emu else np.nan)

    scheduler.step(tot_loss_val[-1])
    bar.set_postfix(train=float(tot_loss_train[-1]),
                    val=float(tot_loss_val[-1]),
                    emu=float(tot_loss_emu[-1]),
                    lam=lam)
```
