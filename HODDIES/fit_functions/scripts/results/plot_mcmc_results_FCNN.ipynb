```python
from pathlib import Path
from typing import Any, Sequence

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.figure import Figure


def plot_chain_deltas(
    consumer: Any,
    chains: Sequence[str | Any] | None = None,
    columns: Sequence[str] | None = None,
    filename: str | Path | None = None,
    figsize: float = 1.0,
    parameter_width: float = 2.2,
    row_height: float = 0.65,
    confidence_intervals: bool = True,
    symmetric_limits: bool = True,
    zero_line: bool = True,
) -> Figure:
    r"""Plot parameter offsets from truth for chains in a ChainConsumer object.

    For every chain and parameter, the plotted quantity is

        delta = best_fit - truth

    The function is external to ChainConsumer and does not modify the package.

    Parameters
    ----------
    consumer:
        A populated ChainConsumer instance containing chains and truths.

    chains:
        Optional chain names or chain objects to plot. If omitted, all loaded
        chains are used.

    columns:
        Optional parameter names to plot. If omitted, every parameter shared
        with the loaded truth information is plotted.

    filename:
        Optional output filename.

    figsize:
        Overall figure scaling factor.

    parameter_width:
        Width allocated to each parameter column before applying ``figsize``.

    row_height:
        Height allocated to each chain before applying ``figsize``.

    confidence_intervals:
        Plot marginalized lower and upper intervals when available.

    symmetric_limits:
        Use x-axis limits symmetric about zero.

    zero_line:
        Draw the vertical delta=0 reference line.

    Returns
    -------
    matplotlib.figure.Figure
        The generated figure.
    """

    def get_loaded_chains(obj: Any) -> list[Any]:
        """Extract all loaded chains from different ChainConsumer versions."""
        loaded = getattr(obj, "chains", None)

        if loaded is None:
            loaded = getattr(obj, "_chains", None)

        if loaded is None:
            raise AttributeError(
                "Could not find chains on the supplied consumer. Expected "
                "`consumer.chains` or `consumer._chains`."
            )

        if isinstance(loaded, dict):
            return list(loaded.values())

        return list(loaded)

    def select_chains(
        loaded_chains: list[Any],
        requested: Sequence[str | Any] | None,
    ) -> list[Any]:
        if requested is None:
            return loaded_chains

        by_name = {
            str(chain.name): chain
            for chain in loaded_chains
        }

        selected = []

        for item in requested:
            if isinstance(item, str):
                if item not in by_name:
                    available = ", ".join(by_name)
                    raise KeyError(
                        f"Unknown chain {item!r}. Available chains: {available}"
                    )

                selected.append(by_name[item])
            else:
                selected.append(item)

        return selected

    def get_loaded_truths(obj: Any) -> list[Any]:
        """Extract loaded Truth objects from the consumer."""
        truths = getattr(obj, "truths", None)

        if truths is None:
            truths = getattr(obj, "_truths", None)

        if truths is None:
            raise AttributeError(
                "Could not find truths on the supplied consumer. Expected "
                "`consumer.truths` or `consumer._truths`."
            )

        if isinstance(truths, dict):
            # This also supports a direct {parameter: truth_value} mapping.
            if all(np.isscalar(value) for value in truths.values()):
                return [truths]

            return list(truths.values())

        return list(truths)

    def make_truth_mapping(truth_objects: list[Any]) -> dict[str, float]:
        result: dict[str, float] = {}

        for truth in truth_objects:
            if isinstance(truth, dict):
                location = truth
            else:
                location = getattr(truth, "location", None)

            if location is None:
                continue

            for parameter, value in location.items():
                if value is not None and parameter not in result:
                    result[str(parameter)] = float(value)

        return result

    def get_data_columns(chain: Any) -> list[str]:
        """Obtain the sampled parameter names from a chain."""
        data_columns = getattr(chain, "data_columns", None)

        if data_columns is not None:
            return [str(column) for column in data_columns]

        samples = getattr(chain, "samples", None)

        if samples is not None and hasattr(samples, "columns"):
            return [str(column) for column in samples.columns]

        raise AttributeError(
            f"Could not determine the parameters in chain "
            f"{getattr(chain, 'name', '<unnamed>')!r}."
        )

    def get_summary(
        obj: Any,
        chain: Any,
        parameter: str,
    ) -> Any | None:
        """Retrieve a marginalized parameter summary."""
        analysis = getattr(obj, "analysis", None)

        if analysis is None:
            raise AttributeError(
                "The supplied consumer does not provide `consumer.analysis`."
            )

        getter = getattr(analysis, "get_parameter_summary", None)

        if getter is None:
            raise AttributeError(
                "`consumer.analysis` does not provide "
                "`get_parameter_summary(chain, parameter)`."
            )

        return getter(chain, parameter)

    def extract_summary_values(
        summary: Any,
    ) -> tuple[float, float | None, float | None] | None:
        """Return center, lower and upper from a summary object."""
        if summary is None:
            return None

        if getattr(summary, "all_none", False):
            return None

        # Standard ChainConsumer summary object.
        center = getattr(summary, "center", None)
        lower = getattr(summary, "lower", None)
        upper = getattr(summary, "upper", None)

        # Allow tuple-like summaries as a fallback.
        if center is None and isinstance(summary, (tuple, list)):
            if len(summary) == 3:
                lower, center, upper = summary
            elif len(summary) == 1:
                center = summary[0]

        if center is None:
            return None

        center_value = float(center)
        lower_value = None if lower is None else float(lower)
        upper_value = None if upper is None else float(upper)

        return center_value, lower_value, upper_value

    def get_chain_colour(chain: Any, index: int) -> Any:
        colour = getattr(chain, "color", None)

        if colour is None:
            return plt.get_cmap("tab10")(index % 10)

        # Some configurations store colours as objects exposing a value.
        return getattr(colour, "value", colour)

    def get_parameter_label(obj: Any, parameter: str) -> str:
        config = getattr(obj, "config", None)

        if config is not None:
            get_label = getattr(config, "get_label", None)

            if callable(get_label):
                return str(get_label(parameter))

        return str(parameter)

    loaded_chains = get_loaded_chains(consumer)
    selected_chains = select_chains(loaded_chains, chains)

    if not selected_chains:
        raise ValueError("No chains were selected.")

    truth_values = make_truth_mapping(get_loaded_truths(consumer))

    if not truth_values:
        raise ValueError(
            "No truth values were found in the supplied consumer."
        )

    if columns is None:
        # Preserve chain parameter order instead of sorting alphabetically.
        candidate_columns = []

        for chain in selected_chains:
            for parameter in get_data_columns(chain):
                if parameter not in candidate_columns:
                    candidate_columns.append(parameter)
    else:
        candidate_columns = [str(parameter) for parameter in columns]

    plotted_columns = [
        parameter
        for parameter in candidate_columns
        if parameter in truth_values
    ]

    if not plotted_columns:
        raise ValueError(
            "None of the requested parameters has a loaded truth value."
        )

    summaries: dict[
        tuple[int, str],
        tuple[float, float | None, float | None],
    ] = {}

    extent_values: dict[str, list[float]] = {
        parameter: [0.0]
        for parameter in plotted_columns
    }

    for chain_index, chain in enumerate(selected_chains):
        chain_columns = set(get_data_columns(chain))

        for parameter in plotted_columns:
            if parameter not in chain_columns:
                continue

            summary = get_summary(consumer, chain, parameter)
            values = extract_summary_values(summary)

            if values is None:
                continue

            best_fit, lower, upper = values
            truth = truth_values[parameter]

            delta_best_fit = best_fit - truth
            delta_lower = None if lower is None else lower - truth
            delta_upper = None if upper is None else upper - truth

            summaries[(chain_index, parameter)] = (
                delta_best_fit,
                delta_lower,
                delta_upper,
            )

            extent_values[parameter].append(delta_best_fit)

            if confidence_intervals:
                if delta_lower is not None:
                    extent_values[parameter].append(delta_lower)

                if delta_upper is not None:
                    extent_values[parameter].append(delta_upper)

    parameter_limits: dict[str, tuple[float, float]] = {}

    for parameter, values in extent_values.items():
        finite_values = np.asarray(values, dtype=float)
        finite_values = finite_values[np.isfinite(finite_values)]

        if finite_values.size == 0:
            parameter_limits[parameter] = (-1.0, 1.0)
            continue

        minimum = float(np.min(finite_values))
        maximum = float(np.max(finite_values))

        if symmetric_limits:
            limit = max(abs(minimum), abs(maximum))

            if np.isclose(limit, 0.0):
                truth_scale = abs(truth_values[parameter])
                limit = max(truth_scale * 0.01, 1.0e-12)

            limit *= 1.15
            parameter_limits[parameter] = (-limit, limit)
        else:
            span = maximum - minimum

            if np.isclose(span, 0.0):
                span = max(abs(maximum) * 0.1, 1.0e-12)

            padding = 0.15 * span
            parameter_limits[parameter] = (
                minimum - padding,
                maximum + padding,
            )

    number_of_chains = len(selected_chains)
    number_of_parameters = len(plotted_columns)

    label_width = max(
        1.2,
        0.12 * max(
            len(str(getattr(chain, "name", "Chain")))
            for chain in selected_chains
        ),
    )

    figure_width = (
        label_width + parameter_width * number_of_parameters
    ) * figsize

    figure_height = max(
        1.5,
        (0.8 + row_height * number_of_chains) * figsize,
    )

    gridspec_kw = {
        "width_ratios": (
            [label_width]
            + [parameter_width] * number_of_parameters
        ),
    }

    fig, axes = plt.subplots(
        nrows=number_of_chains,
        ncols=number_of_parameters + 1,
        figsize=(figure_width, figure_height),
        squeeze=False,
        gridspec_kw=gridspec_kw,
    )

    fig.subplots_adjust(
        left=0.04,
        right=0.98,
        top=0.82,
        bottom=0.18,
        wspace=0.05,
        hspace=0.0,
    )

    for chain_index, chain in enumerate(selected_chains):
        colour = get_chain_colour(chain, chain_index)
        chain_name = str(getattr(chain, "name", f"Chain {chain_index + 1}"))

        name_axis = axes[chain_index, 0]
        name_axis.set_axis_off()
        name_axis.text(
            0.95,
            0.5,
            chain_name,
            transform=name_axis.transAxes,
            horizontalalignment="right",
            verticalalignment="center",
            color=colour,
            fontweight="medium",
        )

        for parameter_index, parameter in enumerate(plotted_columns):
            axis = axes[chain_index, parameter_index + 1]

            axis.set_xlim(parameter_limits[parameter])
            axis.set_ylim(-1.0, 1.0)
            axis.set_yticks([])

            if zero_line:
                axis.axvline(
                    0.0,
                    color="black",
                    linestyle="--",
                    linewidth=1.0,
                    alpha=0.55,
                    zorder=1,
                )

            # if chain_index == 0:
            #     label = get_parameter_label(consumer, parameter)
            #     axis.set_xlabel(
            #         rf"$\Delta\,${label}",
            #         fontsize=11,
            #     )

            if chain_index != number_of_chains - 1:
                axis.tick_params(
                    axis="x",
                    which="both",
                    labelbottom=False,
                )
                axis.spines["bottom"].set_visible(False)
            else:
                label = get_parameter_label(consumer, parameter)
                axis.set_xlabel(
                    rf"$\Delta\,${label}",
                    fontsize=11,
                )

            if chain_index != 0:
                axis.spines["top"].set_visible(False)

            summary = summaries.get((chain_index, parameter))

            if summary is None:
                axis.text(
                    0.5,
                    0.5,
                    "N/A",
                    transform=axis.transAxes,
                    horizontalalignment="center",
                    verticalalignment="center",
                    color="0.6",
                    fontsize=8,
                )
                continue

            delta_best_fit, delta_lower, delta_upper = summary

            if (
                confidence_intervals
                and delta_lower is not None
                and delta_upper is not None
            ):
                lower_error = max(
                    0.0,
                    delta_best_fit - delta_lower,
                )
                upper_error = max(
                    0.0,
                    delta_upper - delta_best_fit,
                )

                axis.errorbar(
                    delta_best_fit,
                    0.0,
                    xerr=np.asarray(
                        [[lower_error], [upper_error]],
                        dtype=float,
                    ),
                    fmt="o",
                    markersize=5,
                    capsize=3,
                    linewidth=1.4,
                    color=colour,
                    markerfacecolor=colour,
                    markeredgecolor=colour,
                    zorder=3,
                )
            else:
                axis.plot(
                    delta_best_fit,
                    0.0,
                    marker="o",
                    markersize=5,
                    color=colour,
                    zorder=3,
                )

    fig.suptitle(
        r"Chain comparison: $\Delta p=p_{\mathrm{bf}}-p_{\mathrm{truth}}$",
        fontsize=12,
    )

    if filename is not None:
        output_path = Path(filename)
        output_path.parent.mkdir(parents=True, exist_ok=True)

        fig.savefig(
            output_path,
            dpi=300,
            bbox_inches="tight",
        )

    return fig
```

```python
"""Histogram offsets from truth using a populated consumer; no ChainConsumer import."""
from pathlib import Path
from collections.abc import Mapping
import warnings

import matplotlib.pyplot as plt
from matplotlib.colors import is_color_like
import numpy as np


def plot_chain_delta_histograms(
    consumer,
    chains=None,
    columns=None,
    filename=None,
    *,
    truths_by_chain=None,
    mode="best_fit",
    bins=20,
    ncols=3,
    figsize=None,
    density=False,
    symmetric_limits=True,
    ddof=0,
    print_stats=True,
):
    r"""Plot delta histograms and return ``(fig, stats)``.

    Parameters
    ----------
    consumer : object
        Populated consumer with chains and truths. No ChainConsumer import or
        package modification is needed. NumPy and Matplotlib are required.
    chains : sequence of names or chain objects, optional
        Defaults to all loaded chains.
    columns : sequence of parameter names, optional
        Defaults to the union of sampled parameters with a loaded truth.
    truths_by_chain : mapping, optional
        {chain_name: {parameter: truth_value}}. Values may also be Truth
        objects exposing .location. Required when chains have different
        truths; matching is by name, never inferred from list order.
        When omitted, loaded consumer truths must be mutually consistent
        and are treated as a shared reference.
    mode : {"best_fit", "posterior"}
        "best_fit": one value per chain, obtained from
        consumer.analysis.get_parameter_summary(chain, parameter).center,
        minus truth. Each chain contributes equally. This matches the center
        used by the supplied plot_chain_deltas function; that center is NOT
        necessarily a maximum-likelihood best fit.
        "posterior": histogram sample minus truth separately for each chain.
        chain.weights (or its weight_column) is used when available. Supply
        chains after any required burn-in removal or thinning.
    bins : positive int
        Same bin edges for all chains within each parameter panel.
    ncols : positive int
        Maximum number of parameter panels per row.
    figsize : (width, height), optional
        Matplotlib figure dimensions in inches.
    density : bool
        Normalize histogram area to one instead of showing counts/weight sums.
    symmetric_limits : bool
        Include zero and use symmetric horizontal limits.
    ddof : {0, 1}
        Best-fit mode: 0 gives population std; 1 gives sample std (NaN for
        fewer than two chains). Posterior mode requires 0 and reports the
        weighted posterior population std, not an error on the mean.
    print_stats : bool
        Also print the statistics; they are always included in panel legends.

    Returns
    -------
    fig : matplotlib.figure.Figure
    stats : dict
        Best-fit: stats[parameter] contains mean, std, n, deltas, chain_names.
        Posterior: stats[parameter][chain_name] contains mean, std, n.
        Missing parameters, missing centers, and empty finite samples are
        skipped with warnings. Conflicting truth values raise an error.
        Nonfinite samples/weights and zero weights are omitted; negative
        finite weights raise an error. Statistics use the original values,
        not histogram bin centers.
    """
    if mode not in {"best_fit", "posterior"}:
        raise ValueError("mode must be 'best_fit' or 'posterior'.")
    if ddof not in (0, 1) or (mode == "posterior" and ddof != 0):
        raise ValueError("ddof must be 0 or 1; posterior mode requires ddof=0.")
    for value, name in ((bins, "bins"), (ncols, "ncols")):
        if not isinstance(value, (int, np.integer)) or value < 1:
            raise ValueError(f"{name} must be a positive integer.")

    def attribute(obj, name):
        value = getattr(obj, name, None)
        return getattr(obj, "_" + name, None) if value is None else value

    loaded = attribute(consumer, "chains")
    if loaded is None:
        raise AttributeError("Expected consumer.chains or consumer._chains.")
    loaded = list(loaded.values()) if isinstance(loaded, Mapping) else list(loaded)
    by_name = {str(c.name): c for c in loaded}
    if isinstance(chains, str):
        chains = [chains]
    selected = loaded if chains is None else [by_name[c] if isinstance(c, str) else c for c in chains]
    if not selected:
        raise ValueError("No chains selected.")
    names = [str(getattr(c, "name", f"Chain {i + 1}")) for i, c in enumerate(selected)]
    if len(set(names)) != len(names):
        raise ValueError("Selected chain names must be unique.")

    def truth_mapping(truth):
        location = truth if isinstance(truth, Mapping) else getattr(truth, "location", None)
        if not isinstance(location, Mapping):
            raise TypeError("Each truth must be a parameter mapping or a Truth object.")
        result = {}
        for p, v in location.items():
            if v is None:
                continue
            v = float(v)
            if not np.isfinite(v):
                raise ValueError(f"Nonfinite truth for {p!r}.")
            result[p] = v
        return result

    if truths_by_chain is not None:
        if not isinstance(truths_by_chain, Mapping):
            raise TypeError("truths_by_chain must be keyed by chain name.")
        missing_names = [name for name in names if name not in truths_by_chain]
        if missing_names:
            raise ValueError(f"Missing truth mappings for chains: {missing_names}")
        chain_truths = {name: truth_mapping(truths_by_chain[name]) for name in names}
    else:
        truths = attribute(consumer, "truths")
        if truths is None:
            raise ValueError("Supply truths_by_chain or load shared truths into the consumer.")
        if isinstance(truths, Mapping):
            truths = [truths] if all(np.isscalar(v) or v is None for v in truths.values()) else list(truths.values())
        shared_truth = {}
        for truth in truths:
            for p, v in truth_mapping(truth).items():
                if p in shared_truth and shared_truth[p] != v:
                    raise ValueError(
                        f"Conflicting loaded truths for {p!r}; supply "
                        "truths_by_chain={chain_name: truth_mapping_or_object}."
                    )
                shared_truth[p] = v
        chain_truths = {name: shared_truth for name in names}

    def parameters(chain):
        cols = getattr(chain, "data_columns", None)
        if cols is not None:
            return list(cols)
        samples = getattr(chain, "samples", None)
        if samples is None or not hasattr(samples, "columns"):
            raise AttributeError(f"Cannot determine parameters for {chain.name!r}.")
        reserved = {getattr(chain, "weight_column", "weight"), getattr(chain, "posterior_column", "log_posterior")}
        return [p for p in samples.columns if p not in reserved]

    available = [parameters(c) for c in selected]
    if isinstance(columns, str):
        columns = [columns]
    if columns is None:
        columns = list(dict.fromkeys(
            p for name, cols in zip(names, available)
            for p in cols if p in chain_truths[name]
        ))
    else:
        columns = list(dict.fromkeys(columns))
    if not columns:
        raise ValueError("No parameters with truth values selected.")
    # Validate each chain/parameter pair before plotting, avoiding silent
    # omissions when a chain-specific truth has not been supplied.
    missing = [(name, p) for name, cols in zip(names, available)
               for p in columns if p in cols and p not in chain_truths[name]]
    if missing:
        raise ValueError(f"Missing truths for (chain, parameter): {missing}")

    data, stats = {}, {}
    for p in columns:
        groups, offsets, used_names = [], [], []
        for i, (chain, name) in enumerate(zip(selected, names)):
            if p not in available[i]:
                warnings.warn(f"Skipping {name!r}, {p!r}: parameter unavailable.", stacklevel=2)
                continue
            if mode == "best_fit":
                summary = consumer.analysis.get_parameter_summary(chain, p)
                center = getattr(summary, "center", None)
                if center is None and isinstance(summary, (tuple, list)):
                    center = summary[1] if len(summary) == 3 else summary[0] if len(summary) == 1 else None
                if center is None or not np.isfinite(center):
                    warnings.warn(f"Skipping {name!r}, {p!r}: no finite center.", stacklevel=2)
                    continue
                offsets.append(float(center) - chain_truths[name][p])
                used_names.append(name)
            else:
                values = np.asarray(chain.samples[p], dtype=float) - chain_truths[name][p]
                weights = getattr(chain, "weights", None)
                if weights is None:
                    weight_column = getattr(chain, "weight_column", "weight")
                    weights = chain.samples[weight_column] if weight_column in chain.samples else np.ones(values.size)
                weights = np.asarray(weights, dtype=float)
                if values.ndim != 1 or weights.shape != values.shape:
                    raise ValueError(f"Sample/weight shape mismatch in {name!r}.")
                if np.any(np.isfinite(weights) & (weights < 0)):
                    raise ValueError(f"Negative weights in {name!r}.")
                valid = np.isfinite(values) & np.isfinite(weights) & (weights > 0)
                if not valid.any():
                    warnings.warn(f"Skipping {name!r}, {p!r}: no usable samples.", stacklevel=2)
                    continue
                values, weights = values[valid], weights[valid]
                normalized = weights / weights.max()
                normalized /= normalized.sum()
                mean = float(np.sum(normalized * values))
                std = float(np.sqrt(np.sum(normalized * (values - mean)**2)))
                color = getattr(chain, "color", None)
                color = getattr(color, "value", color)
                if not is_color_like(color):
                    color = plt.get_cmap("tab10")(i % 10)
                groups.append((name, values, weights, color, mean, std))
                stats.setdefault(p, {})[name] = dict(mean=mean, std=std, n=len(values))
        if mode == "best_fit" and offsets:
            values = np.asarray(offsets)
            mean = float(values.mean())
            std = float(values.std(ddof=ddof)) if len(values) > ddof else float("nan")
            stats[p] = dict(mean=mean, std=std, n=len(values), deltas=values, chain_names=used_names)
            groups = [("Across chains", values, None, "C0", mean, std)]
        if groups:
            data[p] = groups
    if not data:
        raise ValueError("No usable deltas for the selected chains and parameters.")

    width = min(ncols, len(data))
    height = (len(data) + width - 1) // width
    fig, axes = plt.subplots(height, width, squeeze=False, figsize=figsize or (4.4 * width, 3.5 * height), layout="constrained", facecolor="white")
    for ax, (p, groups) in zip(axes.flat, data.items()):
        ax.set_facecolor("white")
        low = min(float(g[1].min()) for g in groups)
        high = max(float(g[1].max()) for g in groups)
        low, high = min(low, 0.0), max(high, 0.0)
        truth_scale = max((abs(t[p]) for t in chain_truths.values() if p in t), default=0.0)
        if symmetric_limits:
            extent = max(abs(low), abs(high)) or max(truth_scale * 0.01, 1e-12)
            low, high = -1.05 * extent, 1.05 * extent
        else:
            padding = 0.05 * (high - low) or max(truth_scale * 0.01, 1e-12)
            low, high = low - padding, high + padding
        edges = np.linspace(low, high, bins + 1)
        for name, values, weights, color, mean, std in groups:
            label = f"{name} (n={len(values)})\nmean={mean:.4g}, std={std:.4g}"
            ax.hist(values, bins=edges, weights=weights, density=density, histtype="stepfilled", alpha=0.22, color=color)
            ax.hist(values, bins=edges, weights=weights, density=density, histtype="step", linewidth=1.5, color=color, label=label)
            ax.axvline(mean, color=color, linestyle=":", linewidth=1.3)
            if print_stats:
                print(f"{p} | {name}: mean={mean:.6g}, std={std:.6g}, n={len(values)}")
        ax.axvline(0, color="black", linestyle="--", linewidth=1)
        ax.set_xlim(low, high)
        ax.set_xlabel(rf"$\Delta\,${p}")
        ax.set_ylabel("Probability density" if density else "Number of chains" if mode == "best_fit" else "Sum of sample weights")
        ax.legend(fontsize=8)
    for ax in list(axes.flat)[len(data):]:
        ax.set_visible(False)
    fig.suptitle("Summary-center offsets from truth" if mode == "best_fit" else "Posterior sample offsets from truth")
    if filename is not None:
        path = Path(filename)
        path.parent.mkdir(parents=True, exist_ok=True)
        fig.savefig(path, dpi=300, bbox_inches="tight", facecolor="white")
    return fig, stats
```
